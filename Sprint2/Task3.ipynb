[Описание задачи: Создать базовый класс Student в C#, который будет представлять информацию о студентах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Student:

    Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
    Методы: GetInfo(): метод для получения информации о студенте в виде строки.

1. Study(): метод для вывода сообщения о том, что студент учится.

2. TakeExam(): метод для вывода сообщения о сдаче экзамена.

Требования к производным классам:

    1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные атрибуты, такие как Специальность (Specialty). Метод Study() должен быть переопределен для добавления информации о специальности при изучении предмета.
    2. Студент магистратуры (MasterStudent): Должен содержать дополнительные атрибуты, такие как Научный руководитель (ScientificAdvisor). Метод TakeExam() должен быть переопределен для добавления информации о научном руководителе при сдаче экзамена.
    3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Тема диссертации (DissertationTopic). Метод Study() должен быть переопределен для добавления информации о теме диссертации при изучении предмета.]

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class Student
{
    public string Name { get; set; }
    public int Age { get; set; }
    public decimal Curs { get; set; }
    
    public Student(string Name, int Age, decimal Curs)
    {
        this.Name = Name;
        this.Age = Age;
        this.Curs = Curs;
    }
    
    public virtual string GetInfo()
    {
        return $"Информация о студенте: {Name}, Полных лет: {Age}, Курс: {Curs}";
    }
    
    public virtual void Study()
    {
        Console.WriteLine($"{Name} - этот студент учится");
    }
    
    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} успешно сдал(а) экзамен");
    }
}

public class BachelorStudent : Student
{
    public string Specialty { get; set; }

    public MasterStudent Nauhruk{get; set; }

    public void Infonauhruk()
    {
        Console.WriteLine($"У студента {Name} научный руководитель: {Nauhruk.Name}");
    }

    
    public BachelorStudent(string Name, int Age, decimal Curs, string Specialty) 
        : base(Name, Age, Curs)
    {
        this.Specialty = Specialty;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Студент учится на специальности: {Specialty} (уровень высшего образования: бакалавриат)";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности '{Specialty}'");
    }
}

public class MasterStudent : BachelorStudent
{
    public string ScientificAdvisor { get; set; }
    
    public MasterStudent(string Name, int Age, decimal Curs, string Specialty, string ScientificAdvisor) 
        : base(Name, Age, Curs, Specialty)
    {
        this.ScientificAdvisor = ScientificAdvisor;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", у студентки научный руководитель: {ScientificAdvisor} (уровень высшего образования: магистратура)";
    }
    
    public override void TakeExam()
    {
        Console.WriteLine($"{Name} успешно сдала экзамен под руководством {ScientificAdvisor}");
    }
}

public class PhDStudent : Student
{
    public string DissertationTopic { get; set; }
    private MasterStudent master;
    
    public PhDStudent(string Name, int Age, int Curs, string DissertationTopic, MasterStudent master) 
        : base(Name, Age, Curs)
    {
        this.DissertationTopic = DissertationTopic;
        this.master = master;
    }
    public void MasterExam()
    {
        master.TakeExam();
    }
    public override string GetInfo()
    {
        return base.GetInfo() + $", у студента тема диссертации: '{DissertationTopic}' ,(уровень высшего образования: аспирантура)";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает тему: '{DissertationTopic}'");
    }
    public void Share(BachelorStudent bachelor)
    {
        Console.WriteLine($"{Name} делится своими исследованиями на тему: {DissertationTopic} со студентом {bachelor.Name}");
    }
}

class Program
{
    public static void Main()
    {
        BachelorStudent bachelor = new BachelorStudent("Вероника Никольская", 18, 1, "Информационная безопасность");
        MasterStudent master = new MasterStudent("Елена Ковальчук", 21, 2, "Прикладная математика и информатика", "Вокина А.Т.");
        PhDStudent aspirant = new PhDStudent("Геннадий Тяжелов", 23, 4, "Разработка приложений", master);

        Student[] students = {bachelor, master, aspirant};
        
        foreach (var student in students)
        {
            Console.WriteLine(student.GetInfo());
            student.Study();
            student.TakeExam();
            Console.WriteLine();
        }

        aspirant.Share(bachelor);
        bachelor.Nauhruk = master;
        bachelor.Infonauhruk();
        aspirant.MasterExam();

    }
}
Program.Main();
```
