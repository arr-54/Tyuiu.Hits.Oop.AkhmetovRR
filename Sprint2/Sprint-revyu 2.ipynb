```polyglot-notebook
public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void Printing()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary} руб.");
    }
}

public class Manager : Employee
{
    public string Department { get; set; }

    public Manager(string name, string position, double salary, string department) 
        : base(name, position, salary)
    {
        Department = department;
    }

    public override void Printing()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary} руб., Отдел: {Department}");
    }
}

public class ProjectManager : Manager
{
    public string ProjectName { get; set; }

    public ProjectManager(string name, string position, double salary, string department, string projectName) 
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
    }

    public override void Printing()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary} руб., Отдел: {Department}, Проект: {ProjectName}");
    }
}

public interface iReportable
{
    string GenerateReport(); 
}

public interface iTeamLeader
{
    string[] GetTeamMembers(); 
}

public class AdvancedProjectManager : Manager, iReportable, iTeamLeader
{
    public string ProjectName { get; set; }
    private string[] teamMembers; 

    public AdvancedProjectManager(string name, string position, double salary, string department, string projectName, string[] team) 
        : base(name, position, salary, department)
    {
        ProjectName = projectName;
        teamMembers = team;
    }

    public override void Printing()
    {
        Console.WriteLine($"Сотрудник: {Name}, Должность: {Position}, Зарплата: {Salary} руб., Отдел: {Department}, Проект: {ProjectName}");
    }

    public string GenerateReport()
    {
        return ($"Отчет по проекту '{ProjectName}'. " +
               $"Менеджер: {Name}. " +
               $"Команда: {teamMembers.Length} человек.");
    }

    public string[] GetTeamMembers()
    {
        return teamMembers;
    }
}

Console.WriteLine("1. Базовый класс Employee:");
Employee employee = new Employee("Руслан Нахимов", "Разработчик", 50000);
employee.Printing();
Console.WriteLine();

Console.WriteLine("2. Класс Manager:");
Manager manager = new Manager("Никита Евгеньев", "Менеджер IT", 70000, "IT отдел");
manager.Printing();
Console.WriteLine();

Console.WriteLine("3. Класс ProjectManager:");
ProjectManager projectManager = new ProjectManager("Андрей Соколов", "Проектный менеджер", 50000, "Разработка", "Новый веб-сайт");
projectManager.Printing();
Console.WriteLine();

Console.WriteLine("4. AdvancedProjectManager:");
string[] team = {"Алексей Якушев", "Александр Меншиков"};
AdvancedProjectManager advancedPM = new AdvancedProjectManager("Елена Ковальчук", "Старший проектный менеджер", 60000, "Разработка ПО", "Мобильное приложение", team);
advancedPM.Printing();
Console.WriteLine();

Console.WriteLine("5. Тестирование интерфейсов:");
string report = advancedPM.GenerateReport();
Console.WriteLine($"Отчет: {report}");
string[] members = advancedPM.GetTeamMembers();
Console.WriteLine("Члены команды:");
foreach (string member in members)
{
    Console.WriteLine($" - {member}");
}
```
