```polyglot-notebook
public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"{Name} - {Price} руб.";
    }

    public string GetInfo(bool detailed)
    {
        if (detailed)
            return $"{Name} - {Price} руб. (детальная информация)";
        else
            return GetInfo();
    }

    public override string ToString()
    {
        return GetInfo();
    }
}

public interface IInstallable
{
    void Install();
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public double Frequency { get; set; }

    public CPU(string name, decimal price, int cores, double frequency) 
        : base(name, price)
    {
        Cores = cores;
        Frequency = frequency;
    }

    public override string GetInfo()
    {
        return $"[CPU] {Name} ({Cores} ядер, {Frequency} ГГц) - {Price} руб.";
    }

    public void Install()
    {
        Console.WriteLine($"Установка процессора: {Name}");
    }

    public void Install(string technician)
    {
        Console.WriteLine($"Установка процессора: {Name} (монтажник: {technician})");
    }

    public override bool Equals(object obj)
    {
        return Equals(obj as CPU);
    }

    public bool Equals(CPU other)
    {
        if (other is null) return false;
        return Name == other.Name && Cores == other.Cores;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Cores);
    }

    public static bool operator ==(CPU cpu1, CPU cpu2)
    {
        if (cpu1 is null) return cpu2 is null;
        return cpu1.Equals(cpu2);
    }

    public static bool operator !=(CPU cpu1, CPU cpu2)
    {
        return !(cpu1 == cpu2);
    }
}

public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public string Type { get; set; }

    public RAM(string name, decimal price, int capacity, string type) 
        : base(name, price)
    {
        Capacity = capacity;
        Type = type;
    }

    public override string GetInfo()
    {
        return $"[RAM] {Name} ({Capacity} ГБ, {Type}) - {Price} руб.";
    }

    public void Install()
    {
        Console.WriteLine($"Установка памяти: {Name}");
    }

    public override bool Equals(object obj)
    {
        return Equals(obj as RAM);
    }

    public bool Equals(RAM other)
    {
        if (other is null) return false;
        return Name == other.Name && Capacity == other.Capacity;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Capacity);
    }

    public static bool operator ==(RAM ram1, RAM ram2)
    {
        if (ram1 is null) return ram2 is null;
        return ram1.Equals(ram2);
    }

    public static bool operator !=(RAM ram1, RAM ram2)
    {
        return !(ram1 == ram2);
    }
}

public class Computer
{
    public string Name { get; set; }
    private List<Component> components = new List<Component>();

    public Computer(string name)
    {
        Name = name;
    }

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент {component.Name} добавлен");
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        if (components.Remove(component))
        {
            Console.WriteLine($"Компонент {component.Name} удален");
        }
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;
        foreach (var component in components)
        {
            total += component.Price;
        }
        return total;
    }

    public void ShowComponents()
    {
        Console.WriteLine($"\nКомпоненты компьютера '{Name}':");
        foreach (var component in components)
        {
            Console.WriteLine($"  {component.GetInfo()}");
        }
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()} руб.");
    }
}

public static class ComputerFactory
{
    public static Computer CreateBasicComputer(string name)
    {
        Computer computer = new Computer(name);
        
        CPU cpu = new CPU("Intel Core i3", 17000, 5, 5.6);
        RAM ram = new RAM("DDR4 8GB", 2000, 6, "DDR4");
        
        computer.AddComponent(cpu);
        computer.AddComponent(ram);
        
        return computer;
    }
}

Computer basicPC = ComputerFactory.CreateBasicComputer("Базовый ПК");
basicPC.ShowComponents();
CPU cpu = new CPU("Intel Core i5", 20000, 6, 4.0);
Console.WriteLine("\nДемонстрация перегрузки методов:");
Console.WriteLine(cpu.GetInfo());         
Console.WriteLine(cpu.GetInfo(true));       
cpu.Install();                             
cpu.Install("Иван");                        

CPU cpu1 = new CPU("Intel Core i5", 20000, 6, 4.0);
CPU cpu2 = new CPU("Intel Core i5", 20000, 6, 4.0);
Console.WriteLine($"\nСравнение процессоров: {(cpu1 == cpu2 ? "одинаковые" : "разные")}");
```
