[Описание задачи: Создать базовый класс Student в C#, который будет представлять информацию о студентах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Student:

    Атрибуты: Имя (Name), Возраст (Age), Курс (Course).
    Методы: GetInfo(): метод для получения информации о студенте в виде строки.

1. Study(): метод для вывода сообщения о том, что студент учится.

2. TakeExam(): метод для вывода сообщения о сдаче экзамена.

Требования к производным классам:

    1. Студент бакалавриата (BachelorStudent): Должен содержать дополнительные атрибуты, такие как Специальность (Specialty). Метод Study() должен быть переопределен для добавления информации о специальности при изучении предмета.
    2. Студент магистратуры (MasterStudent): Должен содержать дополнительные атрибуты, такие как Научный руководитель (ScientificAdvisor). Метод TakeExam() должен быть переопределен для добавления информации о научном руководителе при сдаче экзамена.
    3. Аспирант (PhDStudent) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Тема диссертации (DissertationTopic). Метод Study() должен быть переопределен для добавления информации о теме диссертации при изучении предмета.]

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и прегегрузкой методов, а также generic классы

```
public class Student
{
    private string _name;
    private int _age;
    private int _course;
    
    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }
    
    public int Age
    {
        get { return _age; }
        set 
        { 
            if (value >= 16 && value <= 70)
                _age = value;
        }
    }
    
    public int Course
    {
        get { return _course; }
        set 
        { 
            if (value >= 1 && value <= 6)
                _course = value;
        }
    }
    
    public string Email { get; set; }
    public double GPA { get; set; }
    
    public static int TotalStudents { get; private set; } = 0;
    
    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        TotalStudents++;
    }
    
    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}";
    }
    
    public virtual string GetInfo(bool includeGPA)
    {
        if (includeGPA)
            return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}, GPA: {GPA:F2}";
        else
            return GetInfo();
    }
    
    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится");
    }
    
    public virtual void Study(string subject)
    {
        Console.WriteLine($"{Name} изучает {subject}");
    }
    
    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен");
    }
    
    public virtual void TakeExam(string examName)
    {
        Console.WriteLine($"{Name} сдаёт экзамен по {examName}");
    }
    
    public virtual void CalculateGPA(double[] grades)
    {
        double sum = 0;
        foreach (double grade in grades)
            sum += grade;
        GPA = sum / grades.Length;
        Console.WriteLine($"{Name} имеет GPA: {GPA:F2}");
    }
    
    public void UpdateEmail(string email)
    {
        Email = email;
        Console.WriteLine($"{Name} обновил email на: {email}");
    }
    
    public static void DisplayTotalStudents()
    {
        Console.WriteLine($"Общее количество студентов: {TotalStudents}");
    }
}

// Студент бакалавриата
public class BachelorStudent : Student
{
    private string _specialty;
    
    public string Specialty
    {
        get { return _specialty; }
        set { _specialty = value; }
    }
    
    public string Faculty { get; set; }
    public int Scholarship { get; set; }
    public bool HasDormitory { get; set; }
    
    public BachelorStudent(string name, int age, int course, string specialty) 
        : base(name, age, course)
    {
        Specialty = specialty;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}";
    }
    
    public override string GetInfo(bool includeGPA)
    {
        string info = base.GetInfo(includeGPA);
        return info + $", Специальность: {Specialty}, Стипендия: {Scholarship} руб.";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает предметы по специальности '{Specialty}'");
    }
    
    public void Study(string subject, int hours)
    {
        Console.WriteLine($"{Name} изучает {subject} в течение {hours} часов");
    }
    
    public void DoLabWork()
    {
        Console.WriteLine($"{Name} выполняет лабораторную работу по {Specialty}");
    }
    
    public void ApplyForScholarship()
    {
        Console.WriteLine($"{Name} подает заявку на стипендию");
    }
    
    public void CheckDormitoryStatus()
    {
        Console.WriteLine($"{Name} {(HasDormitory ? "проживает" : "не проживает")} в общежитии");
    }
}

// Студент магистратуры
public class MasterStudent : Student
{
    private string _scientificAdvisor;
    
    public string ScientificAdvisor
    {
        get { return _scientificAdvisor; }
        set { _scientificAdvisor = value; }
    }
    
    public string ResearchTopic { get; set; }
    public string ThesisTitle { get; set; }
    public bool HasPublication { get; set; }
    
    public MasterStudent(string name, int age, int course, string scientificAdvisor) 
        : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Научный руководитель: {ScientificAdvisor}";
    }
    
    public override string GetInfo(bool includeGPA)
    {
        string info = base.GetInfo(includeGPA);
        return info + $", Научный руководитель: {ScientificAdvisor}, Публикации: {(HasPublication ? "есть" : "нет")}";
    }
    
    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен под руководством {ScientificAdvisor}");
    }
    
    public void TakeExam(string examName, bool withAdvisor)
    {
        if (withAdvisor)
            Console.WriteLine($"{Name} сдаёт {examName} с научным руководителем {ScientificAdvisor}");
        else
            Console.WriteLine($"{Name} сдаёт {examName} самостоятельно");
    }
    
    public void WriteThesis()
    {
        Console.WriteLine($"{Name} пишет диссертацию: '{ThesisTitle}'");
    }
    
    public void ConductResearch()
    {
        Console.WriteLine($"{Name} проводит исследование по теме: '{ResearchTopic}'");
    }
    
    public void SubmitArticle()
    {
        HasPublication = true;
        Console.WriteLine($"{Name} подает статью для публикации");
    }
}

// Аспирант
public class PhDStudent : Student
{
    private string _dissertationTopic;
    
    public string DissertationTopic
    {
        get { return _dissertationTopic; }
        set { _dissertationTopic = value; }
    }
    
    public string ResearchGroup { get; set; }
    public int PublicationsCount { get; set; }
    public string DefenseDate { get; set; }
    
    public PhDStudent(string name, int age, int course, string dissertationTopic) 
        : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Тема диссертации: '{DissertationTopic}'";
    }
    
    public override string GetInfo(bool includeGPA)
    {
        string info = base.GetInfo(includeGPA);
        return info + $", Тема: '{DissertationTopic}', Публикации: {PublicationsCount}";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} исследует тему: '{DissertationTopic}'");
    }
    
    public void Study(bool inLab)
    {
        if (inLab)
            Console.WriteLine($"{Name} проводит исследование в лаборатории по теме: '{DissertationTopic}'");
        else
            Console.WriteLine($"{Name} работает над диссертацией в библиотеке");
    }
    
    public void DefendDissertation()
    {
        Console.WriteLine($"{Name} защищает диссертацию на тему: '{DissertationTopic}' {DefenseDate}");
    }
    
    public void PublishArticle()
    {
        PublicationsCount++;
        Console.WriteLine($"{Name} публикует научную статью. Всего публикаций: {PublicationsCount}");
    }
    
    public void AttendConference(string conferenceName)
    {
        Console.WriteLine($"{Name} участвует в конференции: {conferenceName}");
    }
}

// Generic класс для управления студентами
public class UniversityDepartment<T> where T : Student
{
    private List<T> _students = new List<T>();
    
    public void AddStudent(T student)
    {
        _students.Add(student);
        Console.WriteLine($"Добавлен студент: {student.Name}");
    }
    
    public void RemoveStudent(T student)
    {
        _students.Remove(student);
        Console.WriteLine($"Удален студент: {student.Name}");
    }
    
    public void DisplayAllStudents()
    {
        Console.WriteLine("Список всех студентов:");
        foreach (var student in _students)
        {
            Console.WriteLine(student.GetInfo(true));
        }
    }
    
    public void MakeAllStudy()
    {
        Console.WriteLine("Учебный процесс:");
        foreach (var student in _students)
        {
            student.Study();
        }
    }
    
    public void MakeAllTakeExams()
    {
        Console.WriteLine("Экзаменационная сессия:");
        foreach (var student in _students)
        {
            student.TakeExam();
        }
    }
    
    public int Count => _students.Count;
}

BachelorStudent bachelor = new BachelorStudent("Дмитрий Соколов", 20, 2, "Информатика");
MasterStudent master = new MasterStudent("Светлана Иванова", 22, 3, "д.т.н. Волков");
PhDStudent phd = new PhDStudent("Александр Петров", 23, 4, "Нейросети в медицине");

bachelor.Faculty = "Факультет информационных технологий";
bachelor.Scholarship = 1500;
bachelor.HasDormitory = true;

master.ResearchTopic = "Машинное обучение";
master.ThesisTitle = "Алгоритмы глубокого обучения";
master.HasPublication = true;

phd.ResearchGroup = "AI Research Lab";
phd.PublicationsCount = 5;
phd.DefenseDate = "15.12.2024";

bachelor.CalculateGPA(new double[] { 4.5, 4.0, 4.2, 4.7 });
master.CalculateGPA(new double[] { 4.8, 4.9, 4.7, 5.0 });
phd.CalculateGPA(new double[] { 5.0, 5.0, 4.9, 5.0 });

bachelor.UpdateEmail("dmitry.sokolov@edu.ru");
master.UpdateEmail("svetlana.ivanova@edu.ru");
phd.UpdateEmail("alexander.petrov@edu.ru");

Console.WriteLine("Информация о студентах:");
Console.WriteLine(bachelor.GetInfo(true));
Console.WriteLine(master.GetInfo(true));
Console.WriteLine(phd.GetInfo(true));

Console.WriteLine("Учебный процесс:");
bachelor.Study();
bachelor.Study("Программирование");
bachelor.Study("Математика", 3);
master.Study();
phd.Study();
phd.Study(true);

Console.WriteLine("Экзамены:");
bachelor.TakeExam();
bachelor.TakeExam("Программирование");
master.TakeExam();
master.TakeExam("Машинное обучение", true);
phd.TakeExam();

Console.WriteLine("Дополнительная деятельность:");
bachelor.DoLabWork();
bachelor.ApplyForScholarship();
bachelor.CheckDormitoryStatus();

master.WriteThesis();
master.ConductResearch();
master.SubmitArticle();

phd.DefendDissertation();
phd.PublishArticle();
phd.AttendConference("International AI Conference");

Console.WriteLine("Управление студентами через Generic класс:");
UniversityDepartment<Student> department = new UniversityDepartment<Student>();
department.AddStudent(bachelor);
department.AddStudent(master);
department.AddStudent(phd);

department.DisplayAllStudents();
department.MakeAllStudy();
department.MakeAllTakeExams();

Console.WriteLine($"Всего студентов в отделении: {department.Count}");
Student.DisplayTotalStudents();
```
