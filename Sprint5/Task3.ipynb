Создать базовый класс Student в C#, который будет представлять информацию о студентах. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
public delegate void StudentEventHandler(string message);
public delegate void GradeAddedHandler(string studentName, double grade);
public delegate bool StudentFilter(Student student);

public interface IStudent
{
    void Study();
    void TakeExam();
    string GetInfo();
    double CalculateGPA(double[] grades);
}

public interface IResearch
{
    void ConductResearch();
    void PublishArticle();
}

public class Student : IStudent
{
    private string _name;
    private int _age;
    private int _course;
    private List<double> _grades;
    
    // События
    public event StudentEventHandler StudentCreated;
    public event GradeAddedHandler GradeAdded;
    public event StudentEventHandler StatusChanged;
    
    public string Name { get; set; }
    public int Age { get; set; }
    public int Course { get; set; }
    public string Email { get; set; }
    public double GPA { get; set; }
    public string StudentId { get; set; }
    public bool IsActive { get; set; }
    public DateTime EnrollmentDate { get; set; }
    public string PhoneNumber { get; set; }
    public List<string> Achievements { get; set; }
    
    public List<double> Grades => _grades;
    public static int TotalStudents { get; private set; } = 0;
    
    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
        IsActive = true;
        EnrollmentDate = DateTime.Now;
        _grades = new List<double>();
        Achievements = new List<string>();
        TotalStudents++;
        
        StudentCreated?.Invoke($"{name} зачислен в университет");
    }
    
    // Новые методы
    public void AddGrade(double grade)
    {
        if (grade >= 2.0 && grade <= 5.0)
        {
            _grades.Add(grade);
            GradeAdded?.Invoke(Name, grade);
            UpdateGPA();
        }
    }
    
    public void AddAchievement(string achievement)
    {
        Achievements.Add(achievement);
        Console.WriteLine($"{Name} получил достижение: {achievement}");
    }
    
    public void TransferToCourse(int newCourse)
    {
        int oldCourse = Course;
        Course = newCourse;
        Console.WriteLine($"{Name} переведен с {oldCourse} курса на {newCourse} курс");
    }
    
    private void UpdateGPA()
    {
        if (_grades.Any())
        {
            GPA = _grades.Average();
        }
    }
    
    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}, GPA: {GPA:F2}";
    }
    
    void IStudent.Study()
    {
        Console.WriteLine($"{Name} изучает обязательные дисциплины");
    }
    
    void IStudent.TakeExam()
    {
        Console.WriteLine($"{Name} проходит итоговую аттестацию");
    }
    
    double IStudent.CalculateGPA(double[] grades)
    {
        GPA = grades.Average();
        return GPA;
    }
    
    public virtual void Study()
    {
        Console.WriteLine($"{Name} посещает учебные занятия");
    }
    
    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдаёт экзамен по учебному плану");
    }
    
    public void UpdateEmail(string email)
    {
        Email = email;
        Console.WriteLine($"{Name} зарегистрирован с email: {email}");
    }
    
    public void ChangeStatus(bool isActive)
    {
        IsActive = isActive;
        StatusChanged?.Invoke($"{Name} теперь {(isActive ? "активен" : "неактивен")}");
    }
    
    public static void DisplayTotalStudents()
    {
        Console.WriteLine($"Всего студентов в системе: {TotalStudents}");
    }
}

public class BachelorStudent : Student
{
    public string Specialty { get; set; }
    public string Faculty { get; set; }
    public int Scholarship { get; set; }
    public bool HasDormitory { get; set; }
    public string ClubMembership { get; set; }
    public int PracticeHours { get; set; }
    public List<string> ElectiveCourses { get; set; }
    public string SportsSection { get; set; }
    
    public BachelorStudent(string name, int age, int course, string specialty) 
        : base(name, age, course)
    {
        Specialty = specialty;
        ElectiveCourses = new List<string>();
    }
    
    public void AddElectiveCourse(string course)
    {
        ElectiveCourses.Add(course);
        Console.WriteLine($"{Name} записался на факультатив: {course}");
    }
    
    public void JoinSportsSection(string section)
    {
        SportsSection = section;
        Console.WriteLine($"{Name} записался в спортивную секцию: {section}");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Специальность: {Specialty}";
    }
    
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает базовые дисциплины по специальности '{Specialty}'");
    }
    
    public void JoinClub(string clubName)
    {
        ClubMembership = clubName;
        Console.WriteLine($"{Name} участвует в деятельности клуба: {clubName}");
    }
    
    public void CompletePractice(int hours)
    {
        PracticeHours += hours;
        Console.WriteLine($"{Name} завершил практику, общее количество часов: {PracticeHours}");
    }
}

public class MasterStudent : Student, IResearch
{
    public string ScientificAdvisor { get; set; }
    public string ResearchTopic { get; set; }
    public string ThesisTitle { get; set; }
    public bool HasPublication { get; set; }
    public string ResearchGroup { get; set; }
    public int ConferenceCount { get; set; }
    public List<string> ResearchProjects { get; set; }
    public string Laboratory { get; set; }
    
    public MasterStudent(string name, int age, int course, string scientificAdvisor) 
        : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
        ResearchProjects = new List<string>();
    }
    
    public void AddResearchProject(string project)
    {
        ResearchProjects.Add(project);
        Console.WriteLine($"{Name} участвует в исследовательском проекте: {project}");
    }
    
    public void AssignToLaboratory(string laboratory)
    {
        Laboratory = laboratory;
        Console.WriteLine($"{Name} закреплен за лабораторией: {laboratory}");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Руководитель: {ScientificAdvisor}, Тема: {ResearchTopic}";
    }
    
    public void ConductResearch()
    {
        Console.WriteLine($"{Name} проводит научное исследование: '{ResearchTopic}'");
    }
    
    public void PublishArticle()
    {
        HasPublication = true;
        Console.WriteLine($"{Name} опубликовал статью в научном журнале");
    }
    
    public void SubmitThesis()
    {
        Console.WriteLine($"{Name} представляет магистерскую диссертацию: '{ThesisTitle}'");
    }
    
    public void AttendConference(string conferenceName)
    {
        ConferenceCount++;
        Console.WriteLine($"{Name} выступает на конференции: {conferenceName}");
    }
}

public class PhDStudent : Student, IResearch
{
    public string DissertationTopic { get; set; }
    public string ResearchGroup { get; set; }
    public int PublicationsCount { get; set; }
    public string DefenseDate { get; set; }
    public string ResearchFunding { get; set; }
    public bool IsDissertationCompleted { get; set; }
    public List<string> SupervisedStudents { get; set; }
    public string AcademicDegree { get; set; }
    
    public PhDStudent(string name, int age, int course, string dissertationTopic) 
        : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
        SupervisedStudents = new List<string>();
    }
    
    public void AddSupervisedStudent(string studentName)
    {
        SupervisedStudents.Add(studentName);
        Console.WriteLine($"{Name} руководит студентом: {studentName}");
    }
    
    public void AwardAcademicDegree(string degree)
    {
        AcademicDegree = degree;
        Console.WriteLine($"{Name} получил ученую степень: {degree}");
    }
    
    public override string GetInfo()
    {
        return base.GetInfo() + $", Диссертация: '{DissertationTopic}'";
    }
    
    public void ConductResearch()
    {
        Console.WriteLine($"{Name} проводит фундаментальное исследование: '{DissertationTopic}'");
    }
    
    public void PublishArticle()
    {
        PublicationsCount++;
        Console.WriteLine($"{Name} опубликовал научную статью. Всего публикаций: {PublicationsCount}");
    }
    
    public void DefendDissertation()
    {
        IsDissertationCompleted = true;
        Console.WriteLine($"{Name} успешно защитил диссертацию '{DissertationTopic}'");
    }
    
    public void ReceiveGrant(string grantName)
    {
        ResearchFunding = grantName;
        Console.WriteLine($"{Name} получил исследовательский грант: {grantName}");
    }
}

// Обработчики событий
public static class EventHandlers
{
    public static void OnStudentCreated(string message)
    {
        Console.WriteLine($"{message}");
    }
    
    public static void OnGradeAdded(string studentName, double grade)
    {
        Console.WriteLine($"{studentName} получил оценку: {grade}");
    }
}

// Демонстрация работы
BachelorStudent bachelor = new BachelorStudent("Никита Савцов", 20, 2, "Информатика");
MasterStudent master = new MasterStudent("Елена Ковальчук", 20, 2, "д.т.н. Вокин");
PhDStudent phd = new PhDStudent("Александр Николаев", 21, 3, "Нейросети в медицине");

// Подписка на события
bachelor.StudentCreated += EventHandlers.OnStudentCreated;
bachelor.GradeAdded += EventHandlers.OnGradeAdded;
master.StudentCreated += EventHandlers.OnStudentCreated;
master.GradeAdded += EventHandlers.OnGradeAdded;
phd.StudentCreated += EventHandlers.OnStudentCreated;
phd.GradeAdded += EventHandlers.OnGradeAdded;

// Установка свойств
master.ResearchTopic = "Машинное обучение";
master.ThesisTitle = "Алгоритмы глубокого обучения";

phd.ResearchGroup = "Neural Networks Lab";
phd.DefenseDate = "15.12.2024";

Console.WriteLine("Информация о студентах:");
Console.WriteLine(bachelor.GetInfo());
Console.WriteLine(master.GetInfo());
Console.WriteLine(phd.GetInfo());

Console.WriteLine();
Console.WriteLine("Учебная деятельность:");
bachelor.Study();
master.Study();
phd.Study();

Console.WriteLine();
Console.WriteLine("Экзаменационная сессия:");
bachelor.TakeExam();
master.TakeExam();
phd.TakeExam();

Console.WriteLine();
Console.WriteLine("Научная работа:");
master.ConductResearch();
master.PublishArticle();
phd.ConductResearch();
phd.PublishArticle();

Console.WriteLine();
Console.WriteLine("Дополнительная активность:");
bachelor.JoinClub("Программирование");
bachelor.CompletePractice(120);
bachelor.AddElectiveCourse("Веб-разработка");
bachelor.JoinSportsSection("Силовой фитнес");

master.AttendConference("International AI Conference");
master.SubmitThesis();
master.AddResearchProject("Компьютерное зрение");
master.AssignToLaboratory("Лаборатория ИИ");

phd.ReceiveGrant("РНФ №24-456");
phd.DefendDissertation();
phd.AddSupervisedStudent("Николай Егоров");
phd.AwardAcademicDegree("Кандидат наук");

Console.WriteLine();
Console.WriteLine("Успеваемость:");
bachelor.AddGrade(4.5);
bachelor.AddGrade(4.8);
bachelor.AddGrade(4.2);
bachelor.AddAchievement("Призер олимпиады");

master.AddGrade(4.7);
master.AddGrade(4.9);
master.AddGrade(4.6);

phd.AddGrade(5.0);
phd.AddGrade(4.8);
phd.AddGrade(4.9);

Console.WriteLine();
Console.WriteLine("Итоговая информация:");
Console.WriteLine(bachelor.GetInfo());
Console.WriteLine(master.GetInfo());
Console.WriteLine(phd.GetInfo());

Student.DisplayTotalStudents();
```
