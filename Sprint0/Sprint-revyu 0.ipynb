```polyglot-notebook
using System;

public class Plant
{
    public string Name;     
    public string Type;     
    public float Height;

    public Plant(string name, string type, float height)
    {
        this.Name = name;
        this.Type = type;
        this.Height = height;
    }

    public virtual void Blossom()
    {
        Console.WriteLine($"Данное растение {Name} цветет");
    }

    public void Big(float big)
    {
        Height += big;
        Console.WriteLine($"это растение {Name} выросло на {big} см. Новая высота растения: {Height} см");
    }
}

public class Flower : Plant
{
    public string Color;         
    public int Count; 

     public Flower(string name, float height, string color, int count) 
        : base(name, "Цветок", height) 
         {
        Color = color;
        Count = count;
         }

     public override void Blossom()
     {
        Console.WriteLine($"Цветок {Name} распускается и радует нас! Цветок имеет цвет: {Color}, Количество лепестков: {Count}");
     }
     public void Wither()
    {
        Console.WriteLine($"Цветок {Name} начинает увядать.");
    }
}

public class Tree : Plant
{
    public int Age;            
    public string LeavesType;

    public Tree(string name, float height, int age, string leavesType) 
        : base(name, "Дерево", height)
    {
        this.Age = age;
        this.LeavesType = leavesType;
    }

     public override void Blossom()
    {
        Console.WriteLine($"Дерево {Name} покрывается новыми листьями типа: '{LeavesType}'!");
    }

    public void Old(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево {Name} стало старше на {years} лет. Его настоящий возраст: {Age} лет");
    }
}

Console.WriteLine("РАСТЕНИЯ В САДУ\n");

Flower tulpan = new Flower("Тюльпан", 17.4f, "Желтый", 8);
Tree bzv = new Tree("Сосна", 129.8f, 13, "Стройная");

Console.WriteLine($"Создан цветок: {tulpan.Name}, высота: {tulpan.Height} см");
Console.WriteLine($"Создано дерево: {bzv.Name}, высота: {bzv.Height} см");
Console.WriteLine();

Console.WriteLine("2.Рост растений:");
tulpan.Big(5.2f);
bzv.Big(30.5f);
Console.WriteLine();

Console.WriteLine("Уникальность цветения:");

Plant[] plants = { tulpan, bzv };

foreach (var plant in plants)
{
    Console.Write($"{plant.Type} {plant.Name}: ");
    plant.Blossom(); 
}
Console.WriteLine();

tulpan.Wither();
bzv.Old(10);
Console.WriteLine();
```
